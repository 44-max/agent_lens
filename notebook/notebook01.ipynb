```python
import ollama
from pydantic import BaseModel
from typing import List

# 1. Define the structure
class CodeReview(BaseModel):
    summary: str
    issues: List[str]
    severity: str

# 2. Call the Local Model (This creates the 'response' variable)
print("🔍 AgentLens is analyzing code...")
response = ollama.chat(
    model='qwen2.5:3b',
    messages=[
        {'role': 'system', 'content': 'Review the code and respond in JSON.'},
        {'role': 'user', 'content': 'def add(a, b): return a + b'}
    ],
    format=CodeReview.model_json_schema(),
)

# 3. NOW you can define 'review' and 'table'
# We parse the string content back into our CodeReview object
import json
content = json.loads(response['message']['content'])
review = CodeReview(**content) 

table = (
    ("Summary", review.summary),
    ("Issues", "\n".join(review.issues) if review.issues else "None"),
    ("Severity", review.severity) 
)

# 4. Print the result
print("\n--- Final Review Table ---")
for label, value in table:
    print(f"{label:10}: {value}")
```
